# NEXUS-RAG: Evaluation Analysis

RAGAS metric deep-dive and system comparison.

```python
import sys; sys.path.insert(0, "..")
from evaluation.benchmark_runner import BenchmarkRunner
runner = BenchmarkRunner()
df = runner.run()
df.head()
```

```python
import plotly.express as px
metrics = ["faithfulness","answer_relevancy","context_precision","context_recall"]
means = df[metrics].mean()
fig = px.bar(x=metrics, y=means.values, title="Mean RAGAS Scores", range_y=[0,1])
fig.show()
```

```python
# Healing attempts analysis
import plotly.graph_objects as go
fig = go.Figure(go.Bar(x=df["question"].str[:40], y=df["healing_attempts"],
    marker_color="#00d4ff"))
fig.update_layout(title="Healing Attempts per Query", xaxis_tickangle=-45)
fig.show()
```

```python
# Comparison with naive baseline
comparison = runner.compare_with_baseline(df)
print("NEXUS-RAG:", comparison["nexus_rag_mean"])
print("Naive RAG:", comparison["baseline_mean"])
```
