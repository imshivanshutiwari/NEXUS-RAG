# NEXUS-RAG: Corpus EDA

Exploratory analysis of the ingested document corpus.

```python
import sys; sys.path.insert(0, "..")
from data.processors.document_store import DocumentStore
import pandas as pd
store = DocumentStore()
docs = store.list_documents()
print(f"Total documents: {len(docs)}")
```

```python
# Source distribution
from collections import Counter
import plotly.express as px
counts = Counter(d.source for d in docs)
fig = px.bar(x=list(counts.keys()), y=list(counts.values()), title="Documents per Source")
fig.show()
```

```python
# Word count distribution
lengths = [len(d.content.split()) for d in docs]
fig = px.histogram(x=lengths, title="Word Count Distribution", nbins=50)
fig.show()
```

```python
# Top terms TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np
texts = [d.content[:2000] for d in docs[:200]]
vec = TfidfVectorizer(max_features=20, stop_words="english")
vec.fit_transform(texts)
print("Top terms:", vec.get_feature_names_out())
```
